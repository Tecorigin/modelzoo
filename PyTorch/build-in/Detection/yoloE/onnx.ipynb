```python
from ultralytics import YOLO

# 加 dynamic=True 开启动态批尺寸
YOLO('yoloe-11s-seg.pt').export(
    format='onnx',
    imgsz=640,
    opset=13,
    simplify=True,
    dynamic=True   # 这一行很关键
)
```

```python
from ultralytics import YOLO
model = YOLO('yolov12n.onnx')
res = model('bus.jpg', imgsz=640, conf=0.05)  # 降低置信度门限
print(len(res[0].boxes.data), res[0].boxes.data)  
res[0].save('debug_vis.png')  # 可视化结果

```
